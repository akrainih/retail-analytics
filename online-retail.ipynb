```python
rfm = pd.read_csv('/Users/anastasia/Downloads/v_rfm.csv')
```

```python
rfm.head(5)
```

```python
rfm[rfm['segments'] == 'at risk'].describe()
```

```python
at_risk = rfm[rfm['segments'] == 'at risk']['monetary']
champions = rfm[rfm['segments'] == 'champion']['monetary']
```

```python
from scipy import stats
```

```python
stats.shapiro(at_risk)
```

```python
stats.shapiro(champions)
```

```python
stats, p_value = stats.mannwhitneyu(at_risk, champions, alternative='two-sided')
print(f"p_value = {p_value}")
if p_value < 0.05:
    print('Разница между средней суммой покупки статистически значима')
else:
    print('Разница между средней суммой покупки статистически не значима')
```
